辞書内包記

```python
w = ['mon', 'tue','wed']
f = ['coffee', 'milk','water']

d = {}
for x, y in zip(w,f):
  d[x] = y

print(d)
```

```python
d = {x: y for x, y in zip(w,f)}
print(d)
```

集合内包記

```python
s = set()

for i in range(10):
  if i % 2 == 0:
  s.add(i)

print(s)
```

```python
s = {i for i in range(10) if % 2 == 0}
print(s)
```

ジェネレータ内包記

```python
def g():
  for i in range(10):
    yield i

g = g()
print(next(g))
print(next(g))
```

```python
g = (i for i in range(10) if i % 2 == 0)
g2 = tuple(i for i in range(10))
print(type(g))
print(type(g2))
print(next(g))
print(next(g))
print(type(g2))
print(next(g))
print(next(g))
```
